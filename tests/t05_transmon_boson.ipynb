```python
import bohnasa
import matplotlib.pyplot as plt 
import numpy as np
```

```python
N = 5
L = 10
fock = bohnasa.fockStates_fixed_particle_number(N,L)

f = np.zeros(L)
f[0] = N
```

```python
hopping = bohnasa.hopping(fock)
hopping.matrix *= 0.005
interaction = bohnasa.interaction(fock)
interaction.matrix *= -0.25
muVec = np.random.random(L)*0.125
mu = bohnasa.chemical_mu(fock,muVec)

test_op = hopping + interaction + mu


```

```python
ev, es = np.linalg.eigh(test_op.matrix.todense())
```

```python
ev[0]
```

```python
maxOv = np.argmax(np.abs(es[:,-1]))
fock.fockArray[maxOv], es[maxOv,-1]
```
